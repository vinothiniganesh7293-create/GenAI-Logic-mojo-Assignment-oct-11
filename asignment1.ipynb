```python
#59. Simulate model evaluation for 10 predictions. Take actual and predicted values one pair at a time, count correct predictions, and calculate accuracy. Use a function for the evaluation and a loop for the repeated inputs.
def evaluate_model():
    correct = 0

    for i in range(10):
        actual = int(input("Enter actual value: "))
        predicted = int(input("Enter predicted value: "))

        if actual == predicted:
            correct = correct + 1

    accuracy = correct / 10

    return accuracy

result = evaluate_model()

print("Accuracy:", result)
print("Accuracy percentage:", result * 100, "%")
```

```python
#60. Build a mini training simulation using functions, a decorator, a loop, conditions, break, and a lambda function. Print the loss for each epoch, stop when loss reaches 0.10 or lower, and return the final loss. Use only the Python concepts covered in class.
def training_decorator(function):
    def wrapper():
        print("Training started")
        result = function()
        print("Training finished")
        return result
    return wrapper

@training_decorator
def train_model():
    loss = 1.0
    reduction = lambda value: value - 0.15

    for epoch in range(1, 21):
        loss = reduction(loss)

        print("Epoch:", epoch, "Loss:", round(loss, 2))

        if loss <= 0.10:
            break

    return loss

final_loss = train_model()

print("Final loss:", round(final_loss, 2))
```
